## 2. Données

Le fichier `donnees_meteo.csv` contient une valeur par jour du 1er janvier au 10 novembre 2026 (314 jours) :

| Colonne | Description | Unité |
|---|---|---|
| `Date` | Jour de la mesure (AAAA-MM-JJ) | – |
| `Temperature` | Température moyenne journalière | °C |
| `Precipitation` | Cumul de précipitations journalier | mm |

Le fichier doit se trouver dans le **même dossier** que ce notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Chargement du fichier CSV (la colonne Date est convertie en date)
data = pd.read_csv("donnees_meteo.csv", parse_dates=["Date"])
```

## 3. Méthode

2. Exploration : `head()`, `info()`, `describe()`.
3. Calcul des indicateurs : moyenne, maximum, minimum (température) ; moyenne et total (précipitations).
4. Visualisation : courbe de température et diagramme en barres des précipitations.

## 4. Résultats

### 4.1 Statistiques sur la température et les précipitations

```python
temp_moyenne = data["Temperature"].mean()
temp_max = data["Temperature"].max()
temp_min = data["Temperature"].min()

precip_moyenne = data["Precipitation"].mean()
precip_totale = data["Precipitation"].sum()

print(f"Température moyenne : {temp_moyenne:.1f} °C")
print(f"Température maximale : {temp_max:.1f} °C")
print(f"Température minimale : {temp_min:.1f} °C")
print(f"Précipitations moyennes : {precip_moyenne:.2f} mm/jour")
print(f"Précipitations totales : {precip_totale:.1f} mm")
```

### 4.2 Graphique de la température

```python
plt.figure(figsize=(10, 4))
plt.plot(data["Date"], data["Temperature"], color="tab:red", linewidth=1)
plt.title("Évolution de la température journalière (janv.–nov. 2026)")
plt.xlabel("Date")
plt.ylabel("Température (°C)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 4.3 Graphique des précipitations

```python
plt.figure(figsize=(10, 4))
plt.bar(data["Date"], data["Precipitation"], color="tab:blue", width=1.0)
plt.title("Précipitations journalières (janv.–nov. 2026)")
plt.xlabel("Date")
plt.ylabel("Précipitations (mm)")
plt.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Interprétation

Les résultats montrent que la température moyenne est de 18,0 °C, avec un maximum de 30,2 °C et un minimum de 4,3 °C. La température présente une évolution saisonnière, avec des valeurs plus faibles en hiver et plus élevées en été.

Les précipitations sont irrégulières au cours de la période étudiée. La moyenne journalière est de 1,82 mm, tandis que le cumul total atteint 571,2 mm.

Les graphiques permettent de visualiser ces variations et de mieux comprendre l'évolution des paramètres météorologiques.

Il faut rappeler que les données sont simulées et ne représentent pas des observations météorologiques réelles.

## 7. Réponses aux questions de synthèse

### 1. Qu'est-ce qu'un logiciel Open Source ?

Un logiciel Open Source est un logiciel dont le code source est accessible au public. Les utilisateurs peuvent étudier son fonctionnement, le modifier et le partager selon les conditions de sa licence.

### 2. Pourquoi utiliser des logiciels Open Source en sciences de l'environnement ?

Les logiciels Open Source permettent aux chercheurs d'analyser les données environnementales, de créer des graphiques et de partager leurs méthodes. Ils sont accessibles, flexibles et favorisent la collaboration scientifique.

### 3. Quel est le rôle de Git ?

Git est un système de gestion de versions qui permet de suivre les modifications d'un projet, d'enregistrer différentes versions et de revenir à une version précédente si nécessaire.

### 4. Quel est le rôle de GitHub ?

GitHub est une plateforme en ligne qui permet d'héberger des dépôts Git, de partager des projets et de collaborer avec d'autres utilisateurs.

### 5. Qu'est-ce qu'un commit ?

Un commit est un enregistrement d'une modification dans un projet. Il permet de conserver une version précise du travail effectué.

### 6. Pourquoi conserver l'historique d'un projet ?

L'historique permet de suivre les modifications, de retrouver les versions précédentes et de comprendre l'évolution du projet. Il facilite également la collaboration et garantit une meilleure traçabilité scientifique.

### 7. Qu'est-ce qu'un Notebook Jupyter ?

Un Notebook Jupyter est un environnement interactif qui permet d'écrire et d'exécuter du code, d'afficher des résultats et des graphiques, et d'ajouter des explications dans un même document.

### 8. Pourquoi utiliser Markdown dans un Notebook ?

Markdown permet d'organiser le Notebook avec des titres, des explications, des méthodes et des interprétations. Il rend l'analyse plus claire et plus facile à comprendre.

### 9. Qu'est-ce qu'une analyse reproductible ?

Une analyse reproductible est une analyse scientifique qu'un autre utilisateur peut refaire en utilisant les mêmes données, les mêmes méthodes et les mêmes outils, afin d'obtenir des résultats cohérents.

### 10. Quels éléments sont nécessaires pour reproduire une analyse environnementale ?

Pour reproduire une analyse environnementale, il faut disposer des données, du Notebook, des bibliothèques nécessaires, des explications, de la méthode utilisée et des résultats obtenus.
