## 2. Données

Le fichier `donnees_meteo.csv` contient une valeur par jour du 1er janvier au 10 novembre 2026 (314 jours) :

| Colonne | Description | Unité |
|---|---|---|
| `Date` | Jour de la mesure (AAAA-MM-JJ) | – |
| `Temperature` | Température moyenne journalière | °C |
| `Precipitation` | Cumul de précipitations journalier | mm |

Le fichier doit se trouver dans le **même dossier** que ce notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Chargement du fichier CSV (la colonne Date est convertie en date)
data = pd.read_csv("donnees_meteo.csv", parse_dates=["Date"])
```

## 3. Méthode

1. Chargement du CSV avec `pd.read_csv`.
2. Exploration : `head()`, `info()`, `describe()`.
3. Calcul des indicateurs : moyenne, maximum, minimum (température) ; moyenne et total (précipitations).
4. Visualisation : courbe de température et diagramme en barres des précipitations.

## 4. Résultats

### 4.1 Statistiques sur la température et les précipitations

```python
temp_moyenne = data["Temperature"].mean()
temp_max = data["Temperature"].max()
temp_min = data["Temperature"].min()

precip_moyenne = data["Precipitation"].mean()
precip_totale = data["Precipitation"].sum()

print(f"Température moyenne : {temp_moyenne:.1f} °C")
print(f"Température maximale : {temp_max:.1f} °C")
print(f"Température minimale : {temp_min:.1f} °C")
print(f"Précipitations moyennes : {precip_moyenne:.2f} mm/jour")
print(f"Précipitations totales : {precip_totale:.1f} mm")
```

### 4.2 Graphique de la température

```python
plt.figure(figsize=(10, 4))
plt.plot(data["Date"], data["Temperature"], color="tab:red", linewidth=1)
plt.title("Évolution de la température journalière (janv.–nov. 2026)")
plt.xlabel("Date")
plt.ylabel("Température (°C)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 4.3 Graphique des précipitations

```python
plt.figure(figsize=(10, 4))
plt.bar(data["Date"], data["Precipitation"], color="tab:blue", width=1.0)
plt.title("Précipitations journalières (janv.–nov. 2026)")
plt.xlabel("Date")
plt.ylabel("Précipitations (mm)")
plt.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Interprétation

*(À compléter avec vos propres observations après exécution. Pistes :)*

- La température suit un **cycle saisonnier** : minimale en hiver, maximale en été.
- Les précipitations sont **irrégulières** : beaucoup de jours sans pluie et quelques épisodes intenses, surtout en saison froide.
- Les données étant simulées, ces tendances sont volontairement simplifiées et ne permettent pas de conclure sur un climat réel.
